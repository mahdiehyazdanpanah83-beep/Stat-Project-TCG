```python
# بررسی و ممیزی ستون‌ها، نوع داده‌ها، مقادیر گمشده و شناسه‌های اتصال (مانند شناسه بیمار)

import pandas as pd
import os

data_dir = r'E:\Uni\Statistics\Project\data'  

files = [
    'clinical.tsv', 
    'follow_up.tsv', 
    'multiomics.csv',
    'tcga_brca_clinical_core_aligned.tsv',
    'tcga_brca_survival_final.csv'
]

for filename in files:
    file_path = os.path.join(data_dir, filename)
    if os.path.exists(file_path):
        print(f"================ فایل: {filename} ================")
        
        sep = '\t' if filename.endswith('.tsv') else ','
        
        try:
            df = pd.read_csv(file_path, sep=sep, low_memory=False)
            
            print(f"تعداد سطرها: {df.shape[0]} | تعداد ستون‌ها: {df.shape[1]}")
            
            print("\n--- نام ستون‌ها و نوع داده‌ها (Data Types) ---")
            print(df.dtypes)
            
            missing_vals = df.isnull().sum()
            missing_percent = (missing_vals / len(df)) * 100
            missing_df = pd.DataFrame({'Missing Count': missing_vals, 'Missing Percentage (%)': missing_percent})
            print("\n--- مقادیر گمشده در هر ستون ---")
            print(missing_df[missing_df['Missing Count'] > 0])
            
            print("\n--- چند سطر اول داده (Head) ---")
            display(df.head(3))
            
        except Exception as e:
            print(f"خطا در خواندن فایل {filename}: {e}")
            
        print("\n" + "="*60 + "\n")
    else:
        print(f"فایل {filename} در مسیر مورد نظر یافت نشد.\n")
```

```python
import pandas as pd

# ۱. بارگذاری فایل‌ها
clinical_df = pd.read_csv(r'E:\Uni\Statistics\Project\data\clinical.tsv', sep='\t', low_memory=False)
multiomics_df = pd.read_csv(r'E:\Uni\Statistics\Project\data\multiomics.csv', low_memory=False)

print(f"ابعاد فایل بالینی: {clinical_df.shape}")
print(f"ابعاد فایل مولکولی: {multiomics_df.shape}")

# ۲. پیدا کردن ستون مناسب برای شناسه بیمار در فایل بالینی
# معمولاً ستون هایی مثل submitter_id یا case_submitter_id شامل کد TCGA هستند
possible_cols = [c for c in clinical_df.columns if 'submitter' in c or 'barcode' in c or 'patient' in c]
print("\nستون‌های احتمالی شناسه در فایل بالینی:", possible_cols[:5])

# انتخاب ستون مناسب (معمولا submitter_id یا مشابه آن)
clin_id_col = 'cases.submitter_id' if 'cases.submitter_id' in clinical_df.columns else possible_cols[0]

# استخراج شناسه‌ها (کدهای استاندارد TCGA مثل TCGA-XX-XXXX)
clinical_df['patient_id'] = clinical_df[clin_id_col].astype(str).str.extract(r'(TCGA-[A-Z0-9]{2}-[A-Z0-9]{4})')[0]
if clinical_df['patient_id'].isnull().all():
    # اگر الگوی بالا پیدا نشد، از کل مقادیر همان ستون به عنوان شناسه استفاده می‌کنیم
    clinical_df['patient_id'] = clinical_df[clin_id_col].astype(str)

# برای فایل مولکولی
mol_id_col = multiomics_df.columns[0]
multiomics_df['patient_id'] = multiomics_df[mol_id_col].astype(str).str.extract(r'(TCGA-[A-Z0-9]{2}-[A-Z0-9]{4})')[0]
if multiomics_df['patient_id'].isnull().all():
    multiomics_df['patient_id'] = multiomics_df[mol_id_col].astype(str)

# ۳. یکتاسازی در سطح بیمار (Unit of Analysis = Patient)
clinical_unique = clinical_df.drop_duplicates(subset=['patient_id']).copy()
multiomics_unique = multiomics_df.drop_duplicates(subset=['patient_id']).copy()

# ۴. ادغام دو وجه داده (Inner Join)
merged_df = pd.merge(clinical_unique, multiomics_unique, on='patient_id', how='inner')

# ۵. تشکیل جدول جریان داده (Data Flow Table)
data_flow = pd.DataFrame({
    'مرحله / فایل': ['فایل بالینی (Clinical)', 'فایل مولکولی (Multiomics)', 'نمونه نهایی چندوجهی (Merged)'],
    'تعداد کل رکوردهای اولیه': [len(clinical_df), len(multiomics_df), '-'],
    'تعداد بیماران یکتا': [clinical_unique['patient_id'].nunique(), multiomics_unique['patient_id'].nunique(), len(merged_df)]
})

print("\n================ جدول جریان داده (Data Flow Table) ================")
display(data_flow)
```
