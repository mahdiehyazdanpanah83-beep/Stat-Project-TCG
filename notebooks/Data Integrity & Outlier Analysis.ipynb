```python
import pandas as pd
import numpy as np

clinical_df = pd.read_csv(r'E:\Uni\Statistics\Project\data\clinical.tsv', sep='\t', low_memory=False)
multiomics_df = pd.read_csv(r'E:\Uni\Statistics\Project\data\multiomics.csv', low_memory=False)

print("=== ۱. بررسی رکوردهای تکراری ===")
print(f"تعداد رکوردهای تکراری در فایل بالینی: {clinical_df.duplicated().sum()}")
print(f"تعداد رکوردهای تکراری در فایل مولکولی: {multiomics_df.duplicated().sum()}")

print("\n=== ۲. بررسی مقادیر مفقود (متغیرهای کلیدی) ===")
key_cols = ['demographic.age_at_index', 'diagnoses.ajcc_pathologic_stage']
for col in key_cols:
    if col in clinical_df.columns:
        missing_n = clinical_df[col].isnull().sum()
        missing_p = (missing_n / len(clinical_df)) * 100
        print(f"متغیر {col}: تعداد مفقود = {missing_n} ({missing_p:.2f}%)")

print("\n=== ۳. بررسی داده‌های پرت (Outliers) در سن بیمار ===")
if 'demographic.age_at_index' in clinical_df.columns:
    ages = pd.to_numeric(clinical_df['demographic.age_at_index'], errors='coerce').dropna()
    Q1, Q3 = np.percentile(ages, [25, 75])
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = ages[(ages < lower_bound) | (ages > upper_bound)]
    print(f"محدوده نرمال سن: {lower_bound:.1f} تا {upper_bound:.1f}")
    print(f"تعداد داده‌های پرت شناسایی‌شده برای سن: {len(outliers)}")
```
