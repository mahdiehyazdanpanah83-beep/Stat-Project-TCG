```python
import pandas as pd

clinical_df = pd.read_csv(r'E:\Uni\Statistics\Project\data\clinical.tsv', sep='\t', low_memory=False)

# یافتن ستون مرحله تومور یا سن برای بررسی ویژگی‌های نمونه
stage_col = [c for c in clinical_df.columns if 'stage' in c.lower() and ('pathologic' in c.lower() or 'overall' in c.lower())][0]

# محاسبه فراوانی و درصد نسبی زیرگروه‌ها جهت ارزیابی سنجه‌های نمونه
stage_distribution = clinical_df[stage_col].value_counts(dropna=False)
stage_percentage = clinical_df[stage_col].value_counts(normalize=True, dropna=False) * 100

summary_table = pd.DataFrame({
    'فراوانی (تعداد)': stage_distribution,
    'درصد (%)': stage_percentage.round(2)
})

print("=== جدول توزیع فراوانی مراحل تومور در نمونه ===")
print(summary_table)
```
